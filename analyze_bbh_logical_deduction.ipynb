```python
relationships = ["has_position_number", "precedes", "follows"]

system_msg = """
<role>
  You are an expert in natural-language understanding and information extraction.
  Your job is to read a user-provided statement that describes the order of a set of items
  (people, vehicles, birds, fruits, etc.) and convert every directly stated fact
  into relationship triples drawn from this fixed vocabulary:
  "has_position_number", "precedes", "follows".
</role>

<behavior>

  <rule name="List-items">
    Before extracting any relationships, identify and list all of the distinct items, people, vehicles, or entities whose order is being described in the user statement.
    Output their names exactly as they appear in the text (case-insensitive, no quotes or articles), replacing internal spaces with a single underscore.
    List them in the order they appear in the source text, separated by commas, with no extra spaces.
  </rule>

  <rule name="Select-quality">
    Carefully read the user-provided statement and identify the single quality or dimension being used to order the items (e.g., "left to right", "oldest to newest", "cheapest to most expensive", "highest to lowest rank", etc.). You must select and state this quality before any reasoning or extraction.
  </rule>

  <rule name="Absolute-position">
    If a sentence gives an explicit ordinal position (e.g. “third”, “leftmost”, “last”, “second-most expensive”):
      • Determine the correct position number for the entity based on BOTH the ordinal word and the total number of items in the list.
      • For a quality worded "cheapest to most expensive" (where 1 is "cheapest" and N is "most expensive" for N items):
          - "second-most expensive" means position N-1.
          - "third-most expensive" means position N-2, and so on.
          - "k-th most expensive" means position (N + 1 - k).
          - "k-th cheapest" means position k.
      • Always calculate and assign the correct position number based on this logic.
      • Emit (Entity, has_position_number, k) using the computed value.
  </rule>

  <rule name="Relative-order">
    For every comparative statement indicating order between two entities (such as “A is more expensive than B”, “A is newer than B”, “A is to the left of B”):

    1. Identify the ordering direction selected (e.g., "newest to oldest").
    2. Based on this direction, determine which end of the ordering is considered position 1.
       • For example:
         - "cheapest to most expensive" → position 1 is cheapest
         - "newest to oldest" → position 1 is newest
         - "left to right" → position 1 is leftmost
    3. Then, for any comparison (e.g., "A is newer than B"):
       • Determine which entity should come first (closer to position 1).
       • Emit both:
         (FirstEntity, precedes, SecondEntity)
         (SecondEntity, follows, FirstEntity)
    4. Use the semantics of the comparison AND the selected quality to decide the correct direction.

    Examples:
      - If the quality is "newest to oldest", then "A is newer than B" → A precedes B.
      - If the quality is "oldest to newest", then "A is newer than B" → B precedes A.
      - If the quality is "most expensive to cheapest", then "A is more expensive than B" → A precedes B.
      - If the quality is "left to right", then "A is to the right of B" → A follows B.

    Always emit both triples for each comparison.
  </rule>

  <rule name="Entity-spelling">
    Preserve the surface form of each entity exactly as it appears in the text (case-insensitive, no quotes or articles). Replace internal spaces with a single underscore.
  </rule>

  <rule name="No-inference">
    Record only facts stated explicitly. Do not infer unstated positions or transitive relations.
  </rule>

  <rule name="No-duplicates">
    Output each triple at most once.
  </rule>

</behavior>

<format>
  1. First, print exactly this heading on its own line (no extra spaces or text):
     ### LIST_TO_ORDER
  2. On the next line, print a single, comma-separated list of all the items/entities being ordered, with no extra spaces.
  3. Then, print exactly this heading on its own line (no extra spaces or text):
     ### LEFT_TO_RIGHT_ORDER_QUALITY
  4. On the next line, print only the ordering quality you identified (e.g. "left to right", "oldest to newest", "cheapest to most expensive", etc.).
  5. Then, print exactly this heading on its own line (no extra spaces or text):
     ### STEP_BY_STEP_REASONING
  6. Then, write out your step-by-step reasoning, explaining how each triple was derived, based on the selected ordering quality.
  7. After the reasoning, print exactly this heading on its own line (no extra spaces or text):
     ### TRIPLES_START
  8. Immediately after the heading, output one triple per line in this form:
     (Subject, relation, Object)
     with a single space after each comma.
  9. Use integers for has_position_number triples; use other entity names for follows/precedes triples.
 10. List triples in the order the corresponding facts appear in the paragraph.
 11. Do not add blank lines, commentary, or any text after the final triple.
</format>
"""

human_msg = "{statement}"

prompt = ChatPromptTemplate([("system", system_msg), ("human", human_msg)])
```

```python
from pydantic import BaseModel, Field
from typing import List
import re


class GraphEntry(BaseModel):
    """A single relationship in the knowledge graph"""
    entity_1: str = Field(description="First entity in the relationship.")
    entity_2: str = Field(description="Second entity in the relationship.")
    edge: str = Field(description="The relationship between entity_1 and entity_2.")


class EntityEntry(BaseModel):
    """A single entity in the knowledge graph"""
    entity: str = Field(description="The entity.")
    classes: List[str] = Field(description="Classes of the entity.")


class KnowledgeGraph(BaseModel):
    """Graph representation of the statement and the question asked by the user"""

    graph: List[GraphEntry] = Field(
        description="Graph representation of the statement. "
                    "Each entry should define a relationship with 'entity_1', 'entity_2', and 'edge'."
    )


def sanitize_entity_name(name: str) -> str:
    """
    Sanitize entity names to make them valid for URIs.

    Args:
        name: The entity name to sanitize

    Returns:
        Sanitized entity name
    """
    # Replace apostrophes with empty string
    name = name.replace("'s", "s")
    name = name.replace("'", "")

    # Replace spaces with underscores
    name = name.replace(" ", "_")

    # Remove any other problematic characters
    name = re.sub(r'[^\w\-_]', '', name)

    return name


def extract_triples_from_llm_str_output(text: str) -> KnowledgeGraph:
    # 1) Split off everything before "### TRIPLES_START"
    if "### TRIPLES_START" in text:
        _, triples_part = text.split("### TRIPLES_START", 1)
    else:
        # fallback: run on entire text
        triples_part = text

    # 2) Only run the regex on triples_part
    # triple_pattern = r'^\(\s*([A-Za-z]+)\s*,\s*([a-z_]+)\s*,\s*([A-Za-z]+)\s*\)$'
    triple_pattern = r'^\(\s*([\w]+)\s*,\s*([\w]+)\s*,\s*([\w]+)\s*\)$'
    graph_entries = []
    for line in triples_part.splitlines():
        line = line.strip()
        m = re.match(triple_pattern, line)
        if not m:
            continue
        e1, rel, e2 = m.groups()
        graph_entries.append(
            GraphEntry(
                entity_1=sanitize_entity_name(e1),
                entity_2=sanitize_entity_name(e2),
                edge=sanitize_entity_name(rel)
            )
        )

    return KnowledgeGraph(graph=graph_entries)
```

```python
def convert_string_to_triples(triples_string):
    # Use regex to find all patterns like (X, Y, Z)
    pattern = r'\(([^,]+),\s*([^,]+),\s*([^)]+)\)'
    matches = re.findall(pattern, triples_string)

    # Convert matches to list of tuples, stripping whitespace
    result = [(s.strip(), p.strip(), o.strip()) for s, p, o in matches]

    return result

ground_truths.loc[:, 'ground_truth_graph'] = ground_truths['ground_truth_graph'].apply(convert_string_to_triples)

ground_truths
```

```python
def compare_triples(df):
    """
    Compare if all ground truth triples are present in the generated triples.

    Args:
        df: DataFrame containing 'ground_truth_graph' and 'generated_triples' columns

    Returns:
        DataFrame with new 'is_pass_graph' column (1 if passing, 0 if failing)
    """
    def check_if_all_triples_found(row):
        ground_truth = row['ground_truth_graph']
        generated = row['generated_triples']

        # Check if all ground truth triples exist in generated triples
        all_found = all(triple in generated for triple in ground_truth)

        return 1 if all_found else 0

    # Create a copy to avoid SettingWithCopyWarning
    result_df = df.copy()

    # Add the new column
    result_df['is_pass_graph'] = result_df.apply(check_if_all_triples_found, axis=1)

    return result_df

# Apply the function to your merged dataframe
graph_merged_df = compare_triples(merged_df)
graph_merged_df
```

## 3. Enriching Generated Graphs by adding missing 'has_position_number'

```python
def infer_missing_positions(triples, entities):
    from collections import defaultdict, deque
    from itertools import permutations

    # Normalization helper
    def norm(e):
        return e.strip().replace(" ", "_").lower()

    # Normalize all entity names
    entities = [norm(e) for e in entities]
    triples = [(norm(e1), rel, norm(e2)) for e1, rel, e2 in triples]

    # Parse anchors and order constraints
    precedes = defaultdict(list)
    follows = defaultdict(list)
    anchor_pos = {}
    for e1, rel, e2 in triples:
        if rel == "precedes":
            precedes[e1].append(e2)
            follows[e2].append(e1)
        elif rel == "follows":
            follows[e1].append(e2)
            precedes[e2].append(e1)
        elif rel == "has_position_number":
            anchor_pos[e1] = int(e2)

    n = len(entities)
    positions = list(range(1, n + 1))
    used_positions = set(anchor_pos.values())

    # Build directed graph for topological sorting
    indegree = {e: 0 for e in entities}
    for e in entities:
        for succ in precedes[e]:
            indegree[succ] += 1

    # Topological sort to get one valid ordering
    queue = deque([e for e in entities if indegree[e] == 0])
    topo_order = []
    seen = set()
    while queue:
        e = queue.popleft()
        if e in seen:
            continue
        seen.add(e)
        topo_order.append(e)
        for succ in precedes[e]:
            indegree[succ] -= 1
            if indegree[succ] == 0:
                queue.append(succ)
    for e in entities:
        if e not in seen:
            topo_order.append(e)

    # Place anchors in a slot array; None for missing
    slots = [None] * n
    entity_to_slot = {}
    for ent, pos in anchor_pos.items():
        slots[pos - 1] = ent
        entity_to_slot[ent] = pos - 1

    # Fill in the rest (try every permutation that fits)
    unanchored = [e for e in topo_order if e not in anchor_pos]
    unassigned_positions = [i for i in range(n) if slots[i] is None]

    # Brute-force search is feasible for small N; for large N, use backtracking.
    for perm in permutations(unanchored):
        test_slots = slots[:]
        for idx, e in zip(unassigned_positions, perm):
            test_slots[idx] = e
        # Now, verify all constraints
        valid = True
        pos_of = {e: i + 1 for i, e in enumerate(test_slots)}
        for a in entities:
            for b in precedes[a]:
                if pos_of[a] >= pos_of[b]:  # a must come before b
                    valid = False
                    break
            for b in follows[a]:
                if pos_of[a] <= pos_of[b]:  # a must come after b
                    valid = False
                    break
            if not valid:
                break
        if valid:
            # Found a valid assignment!
            final = [(e, "has_position_number", str(pos_of[e])) for e in entities]
            out_triples = [t for t in triples if t[1] != "has_position_number"] + final
            return out_triples
    raise ValueError("No valid assignment possible (conflicting anchors or edges)")
```

```python
# Loop through the generated_triples in the merged_df and apply infer_missing_positions function
enriched_graphs_list = []

for i in range(len(kg_list)):
    generated_graph_object = kg_list[i]
    entities = entities_list[i]
    generated_triples = []
    for entry in generated_graph_object.graph:
        generated_triples.append((entry.entity_1, entry.edge, entry.entity_2))

    print(f"triples :- {generated_triples}")
    print(f"entities :- {entities}")
    try:
        enriched_triples = infer_missing_positions(generated_triples, entities)
    except:
        enriched_triples = generated_triples
    enriched_position_triples = []

    for j, (e1, rel, e2) in enumerate(enriched_triples):
        if rel in ['has_position_number']:
            enriched_position_triples.append((e1, rel, e2))

    enriched_graphs_list.append({
        'question_number': i+1,
        'enriched_triples': enriched_position_triples
    })

    print(f"Enriched Graph for Question {i+1}:")
    for triple in enriched_position_triples:
        print(f"({triple[0]}, {triple[1]}, {triple[2]})")
    print("-" * 40)
```

## 4. Getting the final answers

```python
final_answer_prompt_system = """
<role>
    You are an expert in logical reasoning and information extraction for natural language questions about ordered entities.
    Your job is to analyze a list of position triples that describe how items are ranked or ordered according to a stated quality,
    and then select the correct natural language statement from a list of options that matches the ordering facts.
  </role>

  <behavior>
    <rule name="Map-positions">
      Interpret each triple of the form (item, has_position_number, k) using the provided quality.
      The meaning of position number k depends on the direction of the quality. For example, if the quality is "cheapest to most expensive", position 1 means "cheapest", 2 means "second-cheapest", etc.
    </rule>

    <rule name="Option-matching">
      For each option, compare the natural language statement with the entity's actual position from the triples, as defined by the quality used to order. Only one option will exactly match the information in the triples.
    </rule>

    <rule name="No-assumptions">
      Do not make any assumptions beyond the given position triples and the provided quality. Use only the explicit data.
    </rule>
  </behavior>

  <format>
    1. Produce a detailed, step-by-step reasoning section that:
       - Explains how each triple is interpreted given the quality used to order,
       - Walks through each option, comparing its claim to the relevant position from the triples,
       - Clearly identifies the option that is supported by the triples.
    2. Then print this heading on its own line (exactly as written):
       ### FINAL ANSWER
    3. On the very next line, output ONLY the correct option letter in parentheses, e.g. (F)
    4. End without adding anything after that word.
  </format>
"""

final_answer_prompt_human = """
<quality used to order>
{quality_used_to_order}
</quality used to order>

<position triples>
{position_triples}
</position triples>

<options>
{options}
</options>
"""
```

```python
final_answer_prompt = ChatPromptTemplate(
    [("system", final_answer_prompt_system),
     ("human", final_answer_prompt_human)]
)

final_answers_list = []

final_answer_chain = final_answer_prompt | llm_model | StrOutputParser()

# batch questions to pass to the LLM
final_questions_batch = []

print(len(enriched_graphs_list))
print(len(options))

for i in range(len(enriched_graphs_list)):
    enriched_triples = enriched_graphs_list[i]['enriched_triples']

    final_questions_batch.append({
        "position_triples": "\n".join([f"({e1}, {rel}, {e2})" for e1, rel, e2 in enriched_triples]),
        "options": options[i],
        "quality_used_to_order": ordering_quality_list[i]
    })

i = 0
for batch in final_questions_batch:
    print(f"Batch Question : {i + 1}")
    print(f"Position Triples: {batch['position_triples']}")
    print("\n")
    print(f"Options: {batch['options']}")
    print("\n")
    print(f"Quality Used to Order: {batch['quality_used_to_order']}")
    print("-" * 40)
    i += 1
```

```python
def extract_final_answer(text):
    marker = "### FINAL ANSWER"
    lines = text.splitlines()
    for i, line in enumerate(lines):
        if line.strip() == marker:
            # Return the very next non-empty line (in case there are blank lines)
            for next_line in lines[i+1:]:
                if next_line.strip():
                    return next_line.strip()
    return []

# Create a DataFrame to store the final answers
final_answers_df = pd.DataFrame({
    'question_number': [i + 1 for i in range(len(final_answers_list))],
    'question': questions,
    'options': options,
    'ordering_quality': ordering_quality_list,
    'entities': entities_list,
    'generated_triples': [
        [f"({e1}, {rel}, {e2})" for e1, rel, e2 in kg_list[i].graph] for i in range(len(kg_list))
    ],
    'llm_response': final_answers_list,
    'final_answer': [extract_final_answer(answer) for answer in final_answers_str_list],
    'ground_truth': targets
})

final_answers_df
```

```python
# Add is_correct column to check if the final answer matches the ground truth
def check_correctness(row):
    return 1 if row['final_answer'] == row['ground_truth'] else 0

final_answers_df['is_correct'] = final_answers_df.apply(check_correctness, axis=1)
final_answers_df
```
